# Semiquintic cinquefoils on concentric hyperspheres

This notebook plots the exact semiquintic cinquefoil family

$$z=s^5,\qquad w=s^2,$$

intersected with concentric hyperspheres

$$|z|^2+|w|^2=R^2.$$

Writing $s=\rho e^{it}$ gives

$$\rho^{10}+\rho^4=R^2.$$

For each chosen hypersphere radius $R$, we solve this equation for $\rho$ and stereographically project the resulting knot from $S^3_R\subset\mathbb{R}^4$ to $\mathbb{R}^3$.

**No normalization $R\mapsto 1$ and no energy rescaling are used.**


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import brentq
```

## Geometry and stereographic projection

Identify $\mathbb{C}^2\cong\mathbb{R}^4$ by

$$z=x_1+i x_2,\qquad w=x_3+i x_4.$$

We project from the north pole $(0,0,0,R)$ of $S^3_R$ onto the hyperplane $x_4=0$. The radius $R$ is retained explicitly, so different hyperspheres are not first normalized to the unit sphere.


```python
R_values = [1.00, 1.25, 1.50, 1.75, 2.00]
t = np.linspace(0.0, 2.0*np.pi, 5000, endpoint=True)

def rho_for_R(R):
    """Solve rho^10 + rho^4 = R^2 for the positive rho."""
    f = lambda rho: rho**10 + rho**4 - R**2
    return brentq(f, 0.0, max(2.0, R + 1.0))

def stereographic_project_R4_to_R3(x1, x2, x3, x4, R):
    """Project S^3_R from (0,0,0,R) onto x4=0."""
    denom = R - x4
    return (
        R * x1 / denom,
        R * x2 / denom,
        R * x3 / denom,
    )
```

```python
curves = []

for R in R_values:
    rho = rho_for_R(R)

    z = rho**5 * np.exp(5j*t)
    w = rho**2 * np.exp(2j*t)

    x1, x2 = z.real, z.imag
    x3, x4 = w.real, w.imag

    X, Y, Z = stereographic_project_R4_to_R3(
        x1, x2, x3, x4, R
    )
    curves.append((R, rho, X, Y, Z))

print('Hypersphere radii and corresponding |s|:')
for R, rho, *_ in curves:
    print(f'R = {R:4.2f}   rho = {rho:.10f}')
```

```python
fig = plt.figure(figsize=(10, 10))
ax = fig.add_subplot(111, projection='3d')

for R, rho, X, Y, Z in curves:
    ax.plot(
        X, Y, Z,
        linewidth=2.0,
        label=fr'$R={R:.2f}$  ($\\rho={rho:.4f}$)'
    )

# Equal 3-D data scaling.
all_x = np.concatenate([c[2] for c in curves])
all_y = np.concatenate([c[3] for c in curves])
all_z = np.concatenate([c[4] for c in curves])

mins = np.array([all_x.min(), all_y.min(), all_z.min()])
maxs = np.array([all_x.max(), all_y.max(), all_z.max()])
center = 0.5 * (mins + maxs)
halfspan = 0.52 * np.max(maxs - mins)

ax.set_xlim(center[0] - halfspan, center[0] + halfspan)
ax.set_ylim(center[1] - halfspan, center[1] + halfspan)
ax.set_zlim(center[2] - halfspan, center[2] + halfspan)
ax.set_box_aspect((1, 1, 1))

ax.set_title(
    'Semiquintic cinquefoils on concentric $S^3_R$ hyperspheres\n'
    r'$z=s^5,\; w=s^2,\; \rho^{10}+\rho^4=R^2$'
    ' — raw stereographic projection'
)
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.view_init(elev=24, azim=-57)
ax.legend(loc='upper left', fontsize=9)
ax.grid(True)

plt.show()
```
