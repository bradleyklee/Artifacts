
# Artifact 24 interactive geometry

This view presents only the closed cycles and exact matched levels. Use the
controls to change sampling density and suppress background levels.


```python
from pathlib import Path
import sys

HERE = Path.cwd().resolve()
ROOT = HERE if (HERE / "artifact24").is_dir() else HERE.parent
if not (ROOT / "artifact24").is_dir():
    raise RuntimeError(f"Could not locate project root from {HERE}")
sys.path.insert(0, str(ROOT))

import plotly.io as pio
pio.renderers.default = "notebook"

import ipywidgets as widgets
from IPython.display import display, clear_output
from artifact24.plots import build_interactive_figure

red = widgets.IntSlider(value=12, min=8, max=20, step=1, description="Red levels")
aux = widgets.IntSlider(value=8, min=4, max=14, step=1, description="Aux levels")
samples = widgets.IntSlider(value=360, min=240, max=700, step=40, description="Samples")
background = widgets.Checkbox(value=True, description="Show background levels")
button = widgets.Button(description="Rebuild figure", button_style="primary")
output = widgets.Output()

def rebuild(_=None):
    with output:
        clear_output(wait=True)
        fig = build_interactive_figure(
            red_count=red.value,
            auxiliary_count=aux.value,
            samples=samples.value,
            show_background=background.value,
        )
        fig.show(config={"scrollZoom": True, "displaylogo": False})

button.on_click(rebuild)
display(widgets.VBox([
    widgets.HBox([red, aux]),
    widgets.HBox([samples, background, button]),
    output,
]))
rebuild()
```
